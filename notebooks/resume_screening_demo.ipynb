# Resume Screening Demo

This notebook demonstrates text cleaning, skill extraction, scoring, and ranking for a resume screening system.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

from ml.text_processing import clean_text, tokenize
from ml.skill_extractor import extract_skills, required_skills_from_job
from ml.resume_ranker import rank_candidates
```

## 1. Define a job role and description

```python
role = "Data Scientist"
job_description = """
We need a Data Scientist with Python, SQL, machine learning, statistics,
pandas, scikit learn, data visualization, feature engineering, and model deployment.
"""

print(clean_text(job_description))
print(tokenize(job_description)[:20])
```

## 2. Extract required skills

```python
required = required_skills_from_job(role, job_description)
required
```

## 3. Create sample resumes

```python
resumes = [
    {
        "name": "Aisha Khan",
        "text": "Python data scientist with SQL, pandas, numpy, scikit learn, statistics, feature engineering, and model deployment."
    },
    {
        "name": "Rahul Sharma",
        "text": "Frontend developer skilled in HTML, CSS, JavaScript, React, Git, and API integrations."
    },
    {
        "name": "Meera Patel",
        "text": "Data analyst with SQL, Excel, Tableau, Python, pandas, statistics, communication, and data visualization."
    }
]

for resume in resumes:
    print(resume["name"], extract_skills(resume["text"]))
```

## 4. Rank candidates

```python
results = rank_candidates(role, job_description, resumes)
results
```

## 5. View skill gaps

```python
for candidate in results:
    print(candidate["name"])
    print("Score:", candidate["score"])
    print("Matched:", candidate["matched_skills"])
    print("Missing:", candidate["missing_skills"])
    print()
```
